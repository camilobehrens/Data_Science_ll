# 15. Modelos de Ensamble y Mejora

## 15.1 Introducción a ensamble

### Métodos de Ensambles en Machine Learning

Los métodos de ensamble son técnicas avanzadas en Machine Learning que permite combinar varios modelos con el objetivo de mojorar la precisión y robustez de las predicciones. La idea de estos métodos es que, al combinar los resultados de varios modelos, se puede reducir el error de generalización y mejorar la estabilidad de las predicciones. Existen tres técnicas principales de ensamble: **Bagging**, **Boosting** y **Stacking**.


### Bagging (Bootstrap Aggregating)

Bagging es una técnica de ensamble en la que varios modelos simples se entrenan de manera independiante utilizando diferentes subconjuntos de datos de entrenamiento. Cada modelo realiza una predicción y los resultados se combinan (generalmente promediando o votando) para obtener una predicción final más precisa.

* **Ventajas:** reduce la varianza del modelo, especialmente en conjuntos de datos de alta dimensionalidad.
* **Desventajas:** pérdida de interpretabilidad debido al proceso de promediado; alto consumo de recursos computacionales.

El algoritmo más utilizado que implementa Bagging es *Random Forest*, en el que se entrenan varios árboles de decisión en paralelo y sus predicciones se combinan para mejorar el rendimiento.



### Boosting

Boosting es otra técnica de ensamble en la que los modelos simples se entrenan de forma secuencial, y cada modelo posterior corrige los errores cometidos por el modelo anterior. A diferencia de Bagging, los modelos no son independiantes, sino que se contruyen en función de los resultados previos.

* **Ventajas:** reducción del sesgo y mejora en la precisión predictiva al centrarse en los errores de los modelos anteriores.
* **Desventajas:** mayor sensibilidad al sobreajuste y alto costo computacional debido a su naturaleza secuencial.

Algoritmos populares que implementan Boosting son *AdaBoost*, *Gradient Boosting*, *XGBoost* y *LightGBM*. Estos algoritmos son altamente efectivos y se utilizan comúnmente en competencias de Machine Learning por su capacidad para mejorar la precisión.



### Stacking

Stacking es una técnica menos común en comparación con Bagging y Boosting. En Stacking, varios modelos simples son entrenados y sus predicciones se utilizan como entradas para otro modelo, llamado **meta-modelo**, que aprende a combinar los resultados de los primeros para realizar una predicción final. Aunque Stacking no se cubre en profundidad en este curso, es importante conocer que esta técnica puede ofrecer un rendimiento superior al combinar modelos heterogéneos.



### Conclusión

Los métodos de ensamble son herramientas poderosas que permiten mejorar significativamente la precisión de los modelos de Machine Learning al combinar las fortalezas de diferentes modelos simples. Bagging y Boosting son las técnicas más utilizada debido a su capacidad para reducir la varianza y el sesgo, respectivamente.

## Validación de Modelos

### Técnicas de Validación en Machine Learning

La validación de modelos es un paso crucial en el proceso de Machine Learning, ya que permite evaluar el rendimiento de un modelo y su capacidad para generalizar a nuevos datos. Existen varias técnicas de validación que ayudan a reducir el sesgo y la varianza en las estimaciones de error. Entre las técnicas más comunes se encuentran la *Validación Simple*, *LOOCV* (Leave-One-Out Cross Validation) y *K-Fold Cross Validation*.


### Validación Simple

La Validación Simple es una técnica en la que el conjunto de datos se divide en dos subconjuntos: uno para entrenamiento y otro para validación. El modelo se entrena con los datos de entrenamiento y se evalúa con los datos de validación para obtener una estimación de su rendimiento.

* **Ventajas:** es una técnica sensilla y rápida de implementar, con bajo costo computacional.
* **Desventajas:** la estimación del error puede ser altamente variable, dependiendo de cómo se dividan los datos. Además, esta técnica es propensa al sesgo, ya que utiliza un solo subconjunto para la evaluación.


### LOOCV (Leave-One-Out Cross Validation)

En *LOOCV*, cada observación del conjunto de datos se usa una vez como conjunto de prueba, mientras que el resto de las observaciones se utilizan para entrenar el modelo. Esto proceso se repite tantas veces como observaciones haya en el conjunto de datos, promediando los resultados para obtener una estimación del error.

* **Ventajas:** minimiza el sesgo, ya que se entrena con prácticamente todo el conjunto de datos. Es ideal para pequeños conjuntos de datos.
* **Desventajas:** es computacionalmente costoso, ya que el modelo debe ser entrenado tantas veces como observaciones haya. También puede tener alta variabilidad en los resultados debido a la sensibilidad a paqueños cambios en los datos.



### K-Fold Cross Validation

La validación cruzada K-Fold es una técnica en la que el conjunto de datos se divide en *K* subconjuntos o "folds". Se entrena el modelo *K* veces, cada vez utilizando *K-1* folds para entrenamiento y el fold restante para validación. El error promedio de todos los *K* pliegues se utiliza como estimación del error del modelo.

* **Ventajas:** ofrece un buen equilibrio entre sesgos y varianzas, ya que cada observación se usa tanto para entrenamiento como para la validación. Además, no es tan costoso computacionalmente como LOOCV.
* **Desventajas:** aunque es más eficiente que LOOCV, requiere entrenae el modelo varias veces, lo que puede aumentar el tiempo de ejecución en grandes conjuntos de datos.


### Conclusión

Las técnicas de validación vistas son esenciales para obtener una estimación fiable del rendimiento de un modelo de Machine Learning. La elección de la técnica adecuada depende de la cantidad de datos disponibles y del equilibrio que se busque entre el costo computacional y la precisión de la estimación 

## 15.3 Boosting Models

### Modelos de Boosting en Machine Learning

Los modelos de Boosting son una técnica de ensamble que mejora el rendimiento predictivo de los modelos de Machine Learning al combinar múltiples modelos simples o "débiles" de manera secuencial. A diferencia de otras ttécnicas como el Bagging, en Boosting cada modelo posterior corrige los errores cometidos por el anterior, lo que resulta en una mejora progresiva del rendimiento del modelo.


### Funcionamiento de los Modelos de Boosting

El Boosting se basa en la idea de que los modelos simples, llamados *weak learners* o aprendices débiles, se pueden combinar para formar un modelo fuerte. Cada nuevo modelo se entrena en los errores que cometió el modelo anterior, dándoles más peso a las observaciones mal clasificadas o con alto error. El objetivo es reducir los errores de predicción, centrandose en las áreas problemáticas de los datos.

* **Ventajas:** el boosting reduce el sesgo al hacer que cada nuevo modelo aprenda de los errores del anterior. Esto permite mejorar significativamente la precisión de las predicciones.

* **Desventajas:** el boosting puede ser propenso al sobreajuste, especialmente si no se controlan los hiperparámetros. Además, su naturaleza secuencial lo hace más costoso computacionalmente en comparación con otros métodos.


### Algoritmos Populares de Boosting

Existen varios algoritmos de Boosting que son ampliamente utilizados en la práctica, entre los más comunes se encuentran:

* **AdaBoost:** es uno de los algoritmos de Boosting más sencillos. Funciona asignando pesos a cada observación, incrementando el peso de las observaciones mal clasificadas en cada iteración. Los clasificados individuales se combinan para formar un clasificador fuerte.

* **Gradient Boosting:** este algoritmo optimiza la función objetivo minimizando los errores residuales en cada paso. Es flexible y puede ser utilizado tanto para clasificación como para regresión.

* **XGBoost:** es una versión optimizada del Gradient Boosting que se ha vuelto extremadamente popular debido a su velocidad y precisión. Es capaz de manejar grandes conjuntos de datos y complejidades de forma eficiente.

* **LightGBM:** otro algoritmo basado en Gradient Boosting que utiliza técnicas avanzadas de optimización para aumentar la velocidad de entrenamiento y reducir el consumo de memoria.



### Impacto en la Mejora de Predicciones

El Boosting tiene un gran impacto en la mejora de las predicciones al combinar los puntos fuertes de varios modelos simples y corregir los errores de forma iterativa. Esto permite que el modelo final sea mucho más robusto y preciso con un solo modelo base. Los modelos de Boosting son especialmente efectivos en tareas donde la precisión es crítica, como competiciones de Machine Learning y aplicaciones industriales.


**Conclusión**

Los modelos de Boosting ofrecen una potente técnica para mejorar las predicciones en Machine Learning al combinar modelos simples de manera secuencial y corregir sus errores. Aunque requieren más recursos computacionales y pueden ser sensibles al sobreajuste, su capacidad para mejorar el rendimiento predictivo los convierte en una opción valiosa para tareas complejas y exigentes.






## 15.4 Mejora de Modelos

### Técnicas Avanzadas para Mejorar modelos: Optimización de Hiperparámetros y Regularización.

En el proceso de Machine Learning, es fundamental mejorar los modelos para obtener mejores predicciones y minimizar errores. Dos de las técnicas avanzadas más utilizadas para este fin son la *optimización de hiperparámetros* y la *regularización*. Estas técnicas ayudan a ajustar los modelos para evitar problemas como el sobreajuste y mejorar la precisión en los datos de prueba.


### Optimización de Hiperparámetros

Los hiperparámetros son parámetros que no se aprenden directamente a partir de los datos, sino que son definidos antes del proceso de entrenamiento del modelo. Ejemplos de hiperparámetros incluyen la tasa de aprendizaje, el número de árboles en un algoritmo de ensamble, o la profundidad máxima de un árbol de decisión. La optimización de hiperparámetros tiene como objetivo encontrar la mejor combinación de estos valores para maximizar el rendimiento del modelo.


### Métodos de Optimización de Hiperparámetros:

* **Grid Search:** realiza una busqueda exhaustiva en una cuadricula de combinaciones de hiperparámetros. Prueba todas las combinaciones posibles de hiperparámetros y selecciona la que ofrece el mejor rendimiento en un conjunto de validación.

* **Ramdomized Search:** a diferencia de Grid Search, este método selecciona aleatoriamente combinaciones de hiperparámetros dentro de un rango definido. Es más rápido que Grid Search y permite cubrir más combinaciones con menos pruebas.

* **Bayesian Optimization:** utiliza un enfoque probabilístico para seleccionar los valores
 de los hiperparámetros, buscando la combinación más prometedora con base en resultados previos. Es eficiente y reduce el tiempo de búsqueda, pero requiere una implementación más compleja.

* **Halving Grid Search y Halving Randomized Search:** son versiones optimizadas que comienzan con una pequeña muestra de combinaciones y van reduciendo el número de opciones en cada paso, enfocándose en las más prometedoras.

La optimización de hiperparámetros es crucial para mejorar el rendimiento del modelo sin introducir sobreajustes, lo que permite que el modelo generalice mejor en datos nuevos.


### Regularización

La regularización es una técnica que se utiliza para evitar el sobreajuste al penalizar los modelos demasiado complejos. En ML el sobreajuste ocurre cuando el modelo se ajusta demasiado a los datos de entrenamiento, capturando el ruido en lugar de los patrones generales, lo que resulta en un bajo rendimiento en datos nuevos.


### Tipos de Regularización:

* **Lasso (L1 Regularization):** agrega una penalización proporcional al valor absoluto de los coeficientes del modelo. Lasso tiende a reducir algunos coeficientes a cero, lo que resulta en un modelo más simple y ayuda en la selección de características.

* **Ridge (L2 Regularization):** agrega una penalización proporcional al cuadrado de los coeficientes. Aunque no reduce los coeficientes a cero, tiende a minimizar su magnitud, haciendo que el modelo sea menos complejo y reduciendo el sobreajuste.

* **Elastic Net:** es una combinación de Lasso y Ridge, permitiendo beneficiarse de ambos métodos al penalizar tanto el valor absoluto como el cuadrado de los coeficientes.

La regularización ayuda a que los modelos sean más simples y robustos, lo que mejora su capacidad para generalizar a nuevos datos.



### Conclusión

Las técnicas de optimización de hiperparámetros y regularización son fundamentales para mejorar modelos de Machine Learning. La optimización permite encontrar los mejores ajustes posibles, mientras que la regularización controla la complejidad del modelo, evitando el sobreajuste y mejorando su capacidad de generalización. Implementar estas técnicas correctamente puede marcar una gran diferencia en el rendimiento de los modelos en la práctica.
