# 16. Datathon en Data Science

## 16.1 Introducción a la Datathon

Una *Datathon* es un evento en el que se reúnen personas con conocimientos en el manejo de datos, como científicos de datos, inginieros de datos y analistas de datos, con el fin de trabajar en forma colaborativa y multidisciplinaria. El objetivo principal de Datathon es resolver problemas relevantes para un contexto o industria específica, utilizando datos como base para la toma de decisiones.

Estos eventos suelen tener una duración limitada, que puede variar desde unas pocas horas hasta varios días. Durante este tiempo, los equipos participantes desarrollan soluciones innovadoras a los problemas planteados, aplicando diversas técnicas de análisis de datos, como el análisis exploratorio, la limpieza y transformación de datos (Data Wrangling), la creación y entrenamiento de modelos, y la evaluación de su desempeño.

La importancia de una Datathon radica en su capacidad para fomentar el trabajo en equipo, la creatividad y la resolución rápida de problemas. Además, permite a los participantes practicar y mejorar sus habilidades técnicas en un entorno de competencia y colaboración.

## 16.2 Objetivo y herramientas

### Objetivo de estudio

En este Datathon, el objetivo principal es predecir los precios de las viviendas residenciales en la ciudad de Ames, lowa, EEUU. Para lograr esto, se utilizarán diversas técnicas de ciencias de datos que se han estudiado a lo largo del curso.

El desafío consiste en predecir el precio de venta ("SalePrice") de las viviendas a partir de un conjunto de características que describen tanto las propiedades físicas de las viviendas como las condiciones del entorno. Para cada observación en el conjunto de prueba, los participantes deben generar una predicción del precio de venta. 

La métrica que se empleará para evaluar el rendimiento de los modelos será el Error Cuadrático Medio (RMSE), que permite medir que tan lejos están las predicciones de los valores reales de los precios de viviendas.

En resumen, el objetivo de estudio de esta Datathon es diseñar y optimizar modelos predictivos que sean capaces de estimar los precios de venta de las viviendas en Ames, utilizando un enfoque de regresión sobre datos previamente suministrados.
