# 14. Validación y Problemas en ML

## 14.1 Conceptos Clave

### Sesgo y Varianza

Sesgo y varianza son dos conceptos fundamentales en el análisis de modelos de Machine Learning, ya que determinan el rendimiento y la capacidad de generalización de un modelo. Ambos términos están relacionados con el error en las predicciones y su comprensión es clave para evitar problemas como el sub- o sobreajuste.

### Sesgo (Bias)

El sesgo es el error que se introduce debido a que el modelo es demasiado simple y no capta las relaciones complejas en los datos. Cuando un modelo tiene un sesgo alto, significa que las predicciones están sistemáticamente alejados de los valores reales. Esto suele ocurrir cuando el modelo no se ajusta adecuandamente a los datos reales. Esto suele ocurrir cuando el modelo no se ajusta adecuadamente a los datos de entrenamiento, un problema conocido como underfitting.

### Varianza

La varianza mide la sensibilidad del modelo a pequeños cambios en los datos de entrenamiento. Un modelo con varianza alta tiende a ajustarse demasiado a los datos de entrenamiento, capturando no solo patrones reales, sino también el ruido o variaciones irrelavantes en los datos. Esto puede llevar a un overfitting, donde el modelo funciona muy bien en el conjunto de entrenamiento pero tiene un rendimiento deficiente en el conjunto de prueba.

Un modelo con varianza baja, por otro lado, es más robusto, ya que pequeños cambios en los datos no alteran singnificativamente sus predicciones.


### Bias-Viariance Tradeoff

El Bias-Variance Tradeoff es el equilibrio que debe alcanzarse entre el sesgo y la varianza para minimizar el error total de un modelo. Disminuir el sesgo a menudo aumenta la varianza y veceversa. El objetivo es encontrar un punto de equilibrio donde el modelo no sea demasiado simple ni demasiado complejo, para lograr un buen rendimiento tanto en el conjunto de entrenamiento como en los datos de prueba.

El error total de un modelo se puede descomponer en tres componentes principales:

        Error = Sesgo^2 + Varianza + Error Irreducible

donde sesgo y varianza sin errores reducibles que dependen del modelo, mientras que el error irreducible es un tipo de error que no puede ser eliminado ni mejorado con ningún ajuste del modelo. 

### Sesgo: error y simplicidad

Un modelo con sesgo alto es típicamente un modelo simple que no ha captado adecuadamente las relaciones subyacentes en los datos. Esta simplicidad excesiva provoca un error en todas las fases: en el conjunto de entrenamiento, en la validación y en la prueba. El *underfitting* es el resultado típico de este problema, ya que el medelo no logra aprender de manera efectiva de los datos de entrenamiento.

La principal causa de un sesgo elevado es la elección de un modelo demasiado restrictivo o con pocos parámetros. Por ejemplo, aplicar un modelo lineal a un problema claramente no lineal genera un sesgo alto porque el modelo no tiene la flexibilidad para capturar la verdadera relación entre variables.

Para reducir el sesgo, se pueden emplear modelos más complejos o aumentar el número de parámetros, lo que permitiría al modelo captar mejor las relaciones entre las variables. Sin embargo, esto debe hacerse con cuidado, ya que un aumento excesivo de la complejidad puede dar lugar a un aumento de la varianza, provocando *overfitting*.


### Varianza y su impacto

Un modelo con **varianza alta** es extremadamente sensible a los cambios en los datos de entrenamiento. Esto significa que si el conjunto de entrenamiento se modifica ligeramente, las predicciones del modelo cambiarán significativamente. La alta varianza es indicativa de un *overfitting*, donde el modelo ha aprendido no solo los patrones generales, sino también el ruido y las peculiaridades del conjunto de datos.

El overfitting suele ocurrir cuando el modelo es demasiado complejo o tiene demasiados parámetros en relación con la cantidad de datos disponibles. Un modelo sobreajustado tiene un desempeño excelente en los datos de entrenamiento, pero no generaliza bien a nuevos datos.

Para controlar la varianza, se puede optar por regularizar el modelo, reducir la complejidad o aumentar el número de datos de entrenamiento. Técnicas como la validación cruzada o el uso de métodos de ensamble, como bagging y boosting, tambie´n ayudan a reducir el impacto de la varianza y mejorar la capacidad de generalización del modelo.
