# 14. Validación y Problemas en ML

## 14.1 Conceptos Clave

### Sesgo y Varianza

Sesgo y varianza son dos conceptos fundamentales en el análisis de modelos de Machine Learning, ya que determinan el rendimiento y la capacidad de generalización de un modelo. Ambos términos están relacionados con el error en las predicciones y su comprensión es clave para evitar problemas como el sub- o sobreajuste.

### Sesgo (Bias)

El sesgo es el error que se introduce debido a que el modelo es demasiado simple y no capta las relaciones complejas en los datos. Cuando un modelo tiene un sesgo alto, significa que las predicciones están sistemáticamente alejados de los valores reales. Esto suele ocurrir cuando el modelo no se ajusta adecuandamente a los datos reales. Esto suele ocurrir cuando el modelo no se ajusta adecuadamente a los datos de entrenamiento, un problema conocido como underfitting.

### Varianza

La varianza mide la sensibilidad del modelo a pequeños cambios en los datos de entrenamiento. Un modelo con varianza alta tiende a ajustarse demasiado a los datos de entrenamiento, capturando no solo patrones reales, sino también el ruido o variaciones irrelavantes en los datos. Esto puede llevar a un overfitting, donde el modelo funciona muy bien en el conjunto de entrenamiento pero tiene un rendimiento deficiente en el conjunto de prueba.

Un modelo con varianza baja, por otro lado, es más robusto, ya que pequeños cambios en los datos no alteran singnificativamente sus predicciones.


### Bias-Viariance Tradeoff

El Bias-Variance Tradeoff es el equilibrio que debe alcanzarse entre el sesgo y la varianza para minimizar el error total de un modelo. Disminuir el sesgo a menudo aumenta la varianza y veceversa. El objetivo es encontrar un punto de equilibrio donde el modelo no sea demasiado simple ni demasiado complejo, para lograr un buen rendimiento tanto en el conjunto de entrenamiento como en los datos de prueba.

El error total de un modelo se puede descomponer en tres componentes principales:

        Error = Sesgo^2 + Varianza + Error Irreducible

donde sesgo y varianza sin errores reducibles que dependen del modelo, mientras que el error irreducible es un tipo de error que no puede ser eliminado ni mejorado con ningún ajuste del modelo. 

### Sesgo: error y simplicidad

Un modelo con sesgo alto es típicamente un modelo simple que no ha captado adecuadamente las relaciones subyacentes en los datos. Esta simplicidad excesiva provoca un error en todas las fases: en el conjunto de entrenamiento, en la validación y en la prueba. El *underfitting* es el resultado típico de este problema, ya que el medelo no logra aprender de manera efectiva de los datos de entrenamiento.

La principal causa de un sesgo elevado es la elección de un modelo demasiado restrictivo o con pocos parámetros. Por ejemplo, aplicar un modelo lineal a un problema claramente no lineal genera un sesgo alto porque el modelo no tiene la flexibilidad para capturar la verdadera relación entre variables.

Para reducir el sesgo, se pueden emplear modelos más complejos o aumentar el número de parámetros, lo que permitiría al modelo captar mejor las relaciones entre las variables. Sin embargo, esto debe hacerse con cuidado, ya que un aumento excesivo de la complejidad puede dar lugar a un aumento de la varianza, provocando *overfitting*.


### Varianza y su impacto

Un modelo con **varianza alta** es extremadamente sensible a los cambios en los datos de entrenamiento. Esto significa que si el conjunto de entrenamiento se modifica ligeramente, las predicciones del modelo cambiarán significativamente. La alta varianza es indicativa de un *overfitting*, donde el modelo ha aprendido no solo los patrones generales, sino también el ruido y las peculiaridades del conjunto de datos.

El overfitting suele ocurrir cuando el modelo es demasiado complejo o tiene demasiados parámetros en relación con la cantidad de datos disponibles. Un modelo sobreajustado tiene un desempeño excelente en los datos de entrenamiento, pero no generaliza bien a nuevos datos.

Para controlar la varianza, se puede optar por regularizar el modelo, reducir la complejidad o aumentar el número de datos de entrenamiento. Técnicas como la validación cruzada o el uso de métodos de ensamble, como bagging y boosting, tambie´n ayudan a reducir el impacto de la varianza y mejorar la capacidad de generalización del modelo.

##  14.2 Equilibrio del modelo

### Tradeof Sesgo-Varianza

El *tradeoff sesgo-varinza* es un concepto clave en el desarrollo de modelos de Machine Learning que describe el equilibrio entre dos fuentes de error: sesgo (bias) y varianza.

Encontrar el balance adecuado entre estos dos factores es crucial para obtener un modelo que generalice bien a nuestros datos.

### Sesgo (bias)

El sesgo es el error que ocurre cuando un modelo es demasiado simple y no logra capturar las relaciones subyacentes en los datos. Un modelo con un **sesgo alto** hace suposiciones fuertes y generaliza mal porque no aprende adecuadamente de los datos de entrenamiento, lo que genera un *underfitting*. Este tipo de error afecta tanto los datos de entrenamiento como los de prueba, y resulta en predicciones incorrectas.

### Varianza

La varianza se refiere a la sensibilidad del modelo a las pequeñas variaciones en los datos de entrenamiento. Un modelo con **varianza alta** es demasiado complejo y se ajusta demasiado bien a los datos de entrenamiento, incluyendo el ruido, lo que lleva a un *overfitting*. Este tipo de modelo tiene un rendimiento excelente en el conjunto de entrenamiento pero falla al predecri correctamente en nuevos datos.


### Equilibrio entre sesgo y Varianza

El desafío en el diseño de modelos es encontrar un equilibrio entre el sesgo y la varianza. Si se disminuye el sesgo al hacer el modelo más complejo, es probable que la varianza aumente. Por otro lado, si se reduce la varianza simplificando el modelo, es probable que el sesgo aumente.

Un **sesgo** bajo con **varianza** alta indica que el modelo se ha sobreajustado a los datos de entrenamiento, mientras que un sesgo alto y con varianza baja sugiere que el modelo no está captando patrones importantes en los datos. El **tradeoff sesgo-varianza** implica que mejorar uno de estos errores generalmente provoca un empeoramiento del otro.


### La importancia de encontrar el equilibrio

El objetivo al entrenar un modelo es minimizar el error total, que se puede descomponer entres componentes: sesgo, varianza y error irreductible

        Error = Sesgo^2 + Varianza + Error Irreducible

El mejor modelo es aquel que tiene un balance adecuado entre sesgo y varianza, lo que minimiza el error total. Un modelo con buen equilibrio es lo suficientemente complejo para capturar los patrones reales de los datos (bajo sesgo), pero no tan complejo como para aprender el ruido o las peculiaridades específicas del conjunto de entrenamiento (baja varianza).

Este balance entre sesgo y varianza es lo que se conoce como **modelo robusto**, ya que puede generalizar correctamente en nuevos datos sin caer en over o underfitting.


### Conclusión

El **Tradeoff sesgo-varianza** es una de las consideraciones más importantes en el desarrollo de modelos de Machine Learning. Entender cómo afecta cada uno el error del modelo permite ajustar la complejidad de los modelos para mejorar su capacidad de generalización. El objetivo es encontrar un equilibrio que minimice el error total y permita que el modelo funciones bien tanto en los datos de entrenamiento como en los de prueba.

## 14.3 Problemas y soluciones

### Problemas Comunes en Machine Leraning

En Machine Learning existen dos problemas recurrentes que afectan el rendimiento de los modelos: over y underfitting. Estos problemas surgen por la relación entre la complejidad del modelo y su capacidad para generalizar a nuevos datos. A continuación se explican ambos problemas con ejemplos prácticos.


### Sobreajuste (overfitting)

El sobreajuste ocurre cuando un modelo se ajusta demasiado a los datos de entrenamiento, capturando no solo los patrones, sino también el ruido y las variaciones aleatorias en los datos. Como resultado, el modelo tiene un rendimiento excelente en el conjunto de entrenamiento, pero falla al generalizar con nuevos datos. 

**Ejemplo práctico:** un modelo de clasificación de imágenes de flores podría aprender a identificar detalles irrelevantes, como la resolución específica de las fotos en el conjunto de entrenamiento, en lugar de centrarse en las características que realmente define a las flores. cuando se le presentan imágenes nuevas (con diferente resolución), el modelo falla.

**Causas del overfitting:**

* Modelo excesivamente complejo (demasiadas características o hiperparámetros mal ajustados).
* Conjunto de entrenamiento pequeño o con ruido.

**Soluciones para evitar el sobreajuste**
1. **Regularización:** aplicar técnicas como Lasso (L1) o Ridge (L2) para penalizar la complejidad del modelo.
2. **Aumentar el tamaño de los datos:** más datos pueden ayudar a que el modelo generalice mejor.
3. **Simplificar el modelo:** reducir el número de características o parámetros para que el modelo no se sobreajuste a los datos de entrenamiento.
4. **Usar técnicas de validación cruzada:** para evaluar mejor el rendimiento del modelo con diferentes subconjuntos de datos.


### Subajuste (underfitting)

El subajuste ocurre cuando el modelo es demasiado simple para captar las relaciones subyacentes en los datos. Como resultado el modelo no funciona bien ni en los datos de entrenamiento ni en los de prueba. Esto indica que el modelo no ha aprendido lo suficiente de los datos disponibles.

**Ejemplo práctico:** si se usa un modelo de regresión lineal para predecir una relación no lineal entre los datos, el modelo subestima la complejidad y no logra realizar predicciones precisas.

**Causas del subajuste:**
* Modelo demasiado simple.
* Falta de suficientes características o variables que capturen la estructura de los datos.

**Soluciones para evitar el subajuste:**
1. **Aumentar la complejidad del modelo:** usar un modelo más flexible o aumentar el número de parámetros.
2. **Agregar características relevantes:** incluir más variables o transformar las existentes para que el modelo pueda captar las relaciones entre variables.
3. **Ajustar los hiperparámetros:** optimizar los hiperparámetros del modelo para mejorar su ajuste a los datos.


Ambos problemas, el sobreajuste y el subajuste, son comunes en Machine Learning. El primero surge cuando el modelo es demasiado complejo, y el segundo, cuando es demasiado simple. La clave es encontrar un equilibrio en la complejidad del modelo para que este pueda generalizar bien en datos nuevos, minimizando ambos problemas.


### Validación de modelos

Es un paso  crucial en Machine Learning para evaluar cómo se comporta un modelo en datos no vistos durante el entrenamiento. El objetivo de la validación es garantizar que el modelo no solo funcione bien con los datos de entrenamiento, sino que también generalice adecuadamente a datos nuevos. Existen varias técnicas de validación, siendo las más comunes la **validación cruzada**.


### Validación Simple

La validación simple consiste en dividir los datos en dos subconjuntos; uno para entrenar el modelo y otro para evaluarlo. Aunque es fácil de implementar, esta técnica puede no ser representativa de todo el conjunto de datos, ya que el rendimiento puede variar mucho dependiendo de cómo se haga la división.


### Validación Cruzada

La validación cruzada es una técnica que divide el conjunto de datos en varios subconjuntos o "folds", lo que permite evaluar el modelo múltiples veces, usando diferentes divisiones de los datos. Esto proporciona una mejor estimación del rendimiento del modelo y ayuda a evitar tanto el *overfitting* como el *underfitting*.

**Técnicas comunes de validación cruzada:**

1. **K-fold Cross-Validation:** el conjunto de datos se divide en *k* partes iguales (folds). El modelo se entrena *k* veces, cada vez utilizando uno de los folds para validación y los restantes para entrenamiento. Luego se promedia el error de las *k* iteraciones para obtener una medida más robusta del rendimiento del modelo.

* **Ventajas:** se usa todo el conjunto de datos tanto para entrenamiento y validación, lo que reduce la variabilidad en la estimación del error.
* **Desventajas:** es computacionalmente más costoso que la validación simple.


2. **Stratified k-fold cross-validation:** es una variante de k-fold, donde los folds se crean de manera que cada uno contenga aproximadamente la misma proporción de clases que el conjunto de datos originales. Esto es útil cuando se trabaja con conjuntos de datos desequilibrados (por ejemplo, cuando una clase es mucho más frecuente que otra).

3. **Leave-One-Out Cross-Validation:** cada observación individual se usa como conjunto de prueba, mietras que el resto de los datos se usa para el entrenamiento. Este proceso se repite para cada observación. Aunque esta técnica da una estimación muy precisa, es muy costosa en términos computacionales, ya que el modelo se entrena tantas veces como observaciones haya.


### Beneficios de la validación cruzada

* **Estimación precisa del rendimiento:** al utilizar múltiples divisiones de los datos, se obtiene una evaluación má precisa y representativa del rendimiento del modelo.

* **Evita el overfitting:** al validar el modelo en diferentes subconjuntos, es menos probable que el modelo se ajuste excesivamente a una sola partición de los datos.

* **Uso eficiente de los datos:** permite utilizar todos los datos disponibles para entrenar y validar el modelo de forma iterativa.


La **validación de modelos** es un paso crucial para asegurar que el modelo generalice bien en datos nuevos. Técnicas como la validación cruzada permiten evaluar el rendimiento del modelo en diferentes subconjunto de datos, ayudando a evitar problemas de overfitting y underfitting. Esto permite tomar decisiones informadas sobre ajustes y mejoras en el modelo.
